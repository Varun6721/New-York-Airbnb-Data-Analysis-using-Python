#### EDA
**Task5.Data Analysis**

**Univariate Analysis**

```python
#identifying outliers in price

sns.boxplot(data=df,x='price')
```

```python
#removing outliers in price

data=df[df['price']<2000]
```

```python
#price distribution
sns.histplot(data=df, x='price')
```

```python
sns.boxplot(data=data,x='price')
```

```python
#price distribution after outliers
plt.figure(figsize=(8,5))
sns.histplot(data=data,x='price',bins=100)
plt.title('Price distribution')
plt.ylabel('Frequency')
plt.show()
```

```python
df.dtypes
```

```python
#availability_365 distribution 
plt.figure(figsize=(8,5))
sns.histplot(data=data,x='availability_365')
plt.title('availability_365 distribution')
plt.ylabel('Frequency')
plt.show()
```

```python
data.groupby('neighbourhood_group')['price'].mean()
```

**Feature Engineering**

```python
data['price per bed']=data['price']/data['beds']
```

```python
data.head()
```

```python
#average price per bed
data.groupby('neighbourhood_group')['price per bed'].mean()
```

```python
# average price in every neighbourhood_group
data.groupby('neighbourhood_group')['price'].mean()
```

**Bi Variable Analysis**

One variable in dependency with another variable

```python
data.columns
```

```python
# price dependances on neighbourhood_group
sns.barplot(data=data,x='neighbourhood_group',y='price',hue='room_type')
```

```python
#number of reviews and price relationship

plt.figure(figsize=(8,5))
plt.title('Locality and Review Dependency')
sns.scatterplot(data=data,x='number_of_reviews',y='price',hue='neighbourhood_group')
plt.show()
```

```python
data.dtypes
```

```python
sns.pairplot(data=data,vars=['price','minimum_nights','number_of_reviews','availability_365'],hue='room_type')
```

```python
#Geographical Distributions for Airbnb Listing

plt.figure(figsize=(10,7))
sns.scatterplot(data=data,x='longitude',y='latitude',hue='room_type')
plt.title('Geographical Distributions for Airbnb Listing')
plt.show()
```
